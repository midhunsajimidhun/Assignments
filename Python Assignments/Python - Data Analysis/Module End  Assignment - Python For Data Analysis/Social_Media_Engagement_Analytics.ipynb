# Task 1 — Data Import & Setup

### Import the required libraries 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### NumPy operations

# Task 2 — Data Cleaning

#### 2.1 Identify missing values

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Missing Value Percentage:")
print((df.isna().sum() / len(df)) * 100)
```

### 2.2 Handle missing values

```python
numeric_cols = df.select_dtypes(include=np.number).columns

categorical_cols = df.select_dtypes(include='object').columns

print("Numerical Columns:")
print(numeric_cols)

print("\nCategorical Columns:")
print(categorical_cols)
```

```python
# Numerical missing values
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical missing values
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# Hashtags: missing values are treated as no hashtags
df['hashtags'] = df['hashtags'].fillna('')

# Date values
df['posted_at'] = df['posted_at'].fillna(
    df['posted_at'].mode()[0]
)

print("Missing values handled successfully!")
print(df.isnull().sum())
```

### 2.3 Duplicate handling

```python
print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
```

### 2.4 Standardize categorical values

```python
# Remove unwanted spaces and standardize text
categorical_cols = [
    'gender',
    'country',
    'post_type',
    'post_category',
    'device_type',
    'sentiment'
]

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()

# Standardize gender labels
df['gender'] = df['gender'].replace({
    'm': 'male',
    'f': 'female',
    'others': 'other'
})

# Standardize sentiment labels
df['sentiment'] = df['sentiment'].replace({
    'pos': 'positive',
    'neg': 'negative',
    'neu': 'neutral'
})

print(df['gender'].value_counts())
print(df['sentiment'].value_counts())
```

### 2.5 Correct unrealistic values

```python
print("Negative ages:", (df['age'] < 0).sum())
print("Negative likes:", (df['likes'] < 0).sum())
print("Negative comments:", (df['comments'] < 0).sum())
print("Negative shares:", (df['shares'] < 0).sum())
```

### 2.6 Extract hashtag count

```python
# Handle missing hashtag values
df['hashtags'] = df['hashtags'].fillna('')

# Extract hashtag count
df['hashtag_count'] = df['hashtags'].str.split().str.len()

# Display the result
df[['hashtags', 'hashtag_count']].head(10)
```

### 2.7 Clean sentiment labels

```python
print("Unique Sentiment Values:")
print(df['sentiment'].unique())

df['sentiment'] = df['sentiment'].str.lower().str.strip()

print("\nCleaned Sentiment Values:")
print(df['sentiment'].unique())
```

# Final Insights and Analysis

### 1. Content Performance

#### 1.1 Which post types have the highest engagement?

```python
post_type_engagement = df.groupby('post_type')['engagement_score'].mean().sort_values(ascending=False)
display(post_type_engagement)
```

#### 1.2 Best-performing content category

```python
category_engagement = df.groupby('post_category')['engagement_score'].mean().sort_values(ascending=False)

display(category_engagement)
```

#### 1.3 Which countries have the highest average engagement rate?

```python
country_insights = df.groupby('country').agg({
    'engagement_rate': 'mean',
    'likes': 'mean',
    'impression_count': 'mean'
}).sort_values(by='engagement_rate', ascending=False)

display(country_insights.round(2))

print("Country with Highest Engagement:",
      country_insights['engagement_rate'].idxmax())
```

### 2. User Trends

#### 2.1 How does age affect engagement?

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[12, 18, 25, 35, 45, 55, 65, 100],
    labels=['13-18', '19-25', '26-35', '36-45',
            '46-55', '56-65', '65+']
)

age_insights = df.groupby(
    'age_group',
    observed=False
)['engagement_rate'].mean().sort_values(ascending=False)

display(age_insights)

print("Age Group with Highest Engagement:",
      age_insights.idxmax())
```

#### 2.2 Performance difference for verified accounts

```python
verified_insights = df.groupby('is_verified').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_rate': 'mean',
    'watch_time_sec': 'mean'
})

display(verified_insights.round(2))

print("Verified Account Engagement:",
      verified_insights['engagement_rate'].get(True))

print("Non-Verified Account Engagement:",
      verified_insights['engagement_rate'].get(False))
```

### 3. Behavioral Insights

#### 3.1 Best time of day for impressions

```python
df['day_name'] = df['posted_at'].dt.day_name()

day_insights = df.groupby('day_name')['impression_count'].mean()

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

day_insights = day_insights.reindex(day_order)

display(day_insights)

print("Best Day for Impressions:",
      day_insights.idxmax())

print("Highest Average Impressions:",
      day_insights.max())
```

#### 3.2 Device type impact on watch time

```python
device_watch_time = df.groupby('device_type')['watch_time_sec'].mean().sort_values(ascending=False)

display(device_watch_time)
```

### 4. Sentiment Analysis

#### 4.1 Which sentiment performs best?

```python
sentiment_engagement = df.groupby('sentiment')['engagement_score'].mean().sort_values(ascending=False)

display(sentiment_engagement)
```

#### 4.2 Behaviour of Negative and Neutral Posts

```python
sentiment_behavior = df[
    df['sentiment'].isin(['negative', 'neutral'])
].groupby('sentiment')[
    ['likes', 'comments', 'shares', 'watch_time_sec', 'engagement_score']
].mean()

display(sentiment_behavior)
```
