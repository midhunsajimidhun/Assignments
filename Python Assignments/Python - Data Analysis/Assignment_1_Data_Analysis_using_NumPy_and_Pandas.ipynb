```python
import pandas as pd
import numpy as np
```

```python
# 1. Create a 1D NumPy Array:
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

```python
# 2. Inspection and Properties:
print("Shape :",temperatures_w1.shape)
print("Data Type :",temperatures_w1.dtype)
print("Number Of Elements :",temperatures_w1.size)
```

```python
# 3. Array Operations:
# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 35
print("Fahrenheit :",temperatures_f)
```

```python
# maximum, minimum, and mean temperatures for the week.
print("Maximum Temperature For The Week :",np.max(temperatures_w1))
print("Minimum Temperature For The Week :",np.min(temperatures_w1))
print("Mean Temperature For The Week :",np.mean(temperatures_w1))
```

```python
# 4. Array Slicing and Indexing:
# Extract the temperatures for the first three days of the week.
print("The temperatures for the first three days of the week :",temperatures_w1[:3])
# Extract the temperatures for the weekend (last two days).
print("The temperatures for the weekend (last two days) :",temperatures_w1[-2:])
# Extract the temperatures for the middle three days of the week.
print("The temperatures for the middle three days of the week :",temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array:
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print("Dimensions:", temperatures.ndim)
```

```python
# 6. Inspect and Slice the 2D Array:
print("Shape :",temperatures.shape)
print("Data Type :",temperatures.dtype)
print("Number Of Elements :",temperatures.size)
```

```python
# Extract the temperatures for each week and the weekends (last two days) for both weeks.
# Extract Week 1
week1 = temperatures[0]
print("Week 1 Temperatures:", week1)

# Extract Week 2
week2 = temperatures[1]
print("Week 2 Temperatures:", week2)

# Extract weekend temperatures for both weeks
weekend = temperatures[:, -2:]
print("Weekend Temperatures:")
print(weekend)
```

# Pandas Series:

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
print("1st Rank Mark:", marks[0])
```

```python
print("Top 3 Ranks :")
print(marks.loc[["Rank1","Rank2","Rank3"]])
```

```python
print("3rd rank student :",marks.iloc[2])
```

```python
print("Marks Greater Than 90 :")
print(marks[marks>90])
```

```python
# Manipulating Series:
# Modify the mark of the student with the 1st rank to 100.
marks["Rank1"] = 100
marks
```

```python
# Remove the entry corresponding to the last rank from the Series.
marks = marks.drop('Rank5')
print(marks)
```

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
# Pandas DataFrame:
# 1. Creating Pandas DataFrame:
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
# 2. Data Exploration:
# Display DataFrame
print("Transactions DataFrame:")
print(transactions)

# First 5 rows
print("\nFirst 5 Rows:")
print(transactions.head())

# Last 5 rows
print("\nLast 5 Rows:")
print(transactions.tail())

# Shape
print("\nShape:", transactions.shape)

# Column names
print("\nColumn Names:")
print(transactions.columns)

# Data types
print("\nData Types:")
print(transactions.dtypes)
```

```python
# Display only the 'ProductCategory' and 'Amount' columns.
transactions[['ProductCategory','Amount']]
```

```python
# Retrieve the last 3 columns of the DataFrame using iloc or loc.
print(transactions.loc[:, ['ProductCategory', 'Region', 'Amount']])
```

```python
# Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
transactions[(transactions['Region'] == 'North') & (transactions['Amount']>200)]
```

```python
print(transactions['ProductCategory'].value_counts())
```

```python
print(transactions['Region'].unique())
```

```python
region_mean = transactions.groupby('Region')['Amount'].mean()

print(region_mean)
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions)
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
# Remove the row with TransactionID 109.
transactions.drop(8,axis=0)
```

```python
# Delete the 'Discount' column from the DataFrame.
transactions.drop('Discount',axis=1)
```
