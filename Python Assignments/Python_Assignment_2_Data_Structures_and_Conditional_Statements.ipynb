### Dictionary (Creation, Modification and Access):

* a. Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).

```python
student_marks = {
    "Arun": 85,
    "Priya": 92,
    "Rahul": 78,
    "Sneha": 88,
    "Karthik": 95
}

print(student_marks)
```

### Sets (Operations):

##### a. Create a set called my_set with following values:
##### ['a','e','i','o','u','a','a','i']
##### Analyse the output and provide explanation for the same.
##### b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.
##### c. Create two sets:

* set1 with values: {1, 3, 5, 7, 9}
* set2 with values: {2, 3, 5, 8, 10}

##### d. Compute and print the union and intersection of set1 and set2.

```python
my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)
```

A set in Python is a collection of unique elements.

Duplicate elements are automatically removed when a set is created.

In the given set, 'a' appears three times and 'i' appears twice.

The duplicate occurrences of 'a' and 'i' are removed, so each vowel appears only once.

Sets are unordered, meaning their elements do not have a fixed position or guaranteed output order.

```python
# Attempt to change the value at index 4
my_set[4] = 's'

print(my_set)
```

* Sets are not indexed: Unlike lists and tuples, sets do not support accessing elements using indexes such as my_set[4].

* Sets are mutable: You can add or remove elements, but you cannot modify an element directly using an index.

```python
## Correct way to replace an element
my_set.remove('u')
my_set.add('s')

print(my_set)
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
```

```python
set1.union(set2)
```

```python
set1.intersection(set2)
```

#### Performance Category Program:

* 1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
* 2. Find the performance category based on the input score using following criteria:
    * a. Above Average: Score greater than 7
    * b. Average: Score between 4 and 7(both inclusive)
    * c. Below Average: Score lesser than 4

```python
score = int(input("Enter Your Score (0 To 10) :"))
if score > 7 :
    print("Above Average: Excellent work! Keep up the great performance.")
elif score >= 4 :
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else :
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
